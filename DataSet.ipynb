```python
url = "https://midocean.edu.km/guide/"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/139.0.0.0 Safari/537.36"
}

response = requests.get(url, headers=headers, timeout=30)

print("Status Code:", response.status_code)
```

```python
# Arabic Text Normalization - Revised

import re

def normalize_arabic(text):
    # إزالة التشكيل والحركات
    text = re.sub(r'[\u0617-\u061A\u064B-\u065F\u0670\u06D6-\u06ED]', '', text)

    # توحيد أشكال الألف
    text = re.sub(r'[إأآٱ]', 'ا', text)

    # توحيد الألف المقصورة إلى الياء
    text = text.replace('ى', 'ي')

    # إزالة التطويل
    text = text.replace('ـ', '')

    # توحيد المسافات
    text = re.sub(r'\s+', ' ', text).strip()

    return text


# إنشاء النص المنظم مع الاحتفاظ بالنص الأصلي
for doc in documents:
    doc["normalized_text"] = normalize_arabic(doc["content"])

print("Arabic normalization completed!")
```

```python
# Section-based Chunking

import re

def create_chunks(text, min_chunk_length=100, max_chunk_length=1200):
    lines = [line.strip() for line in text.split("\n") if line.strip()]

    chunks = []
    current_chunk = ""

    for line in lines:

        # إذا كان السطر عنوانًا قصيرًا
        is_heading = (
            len(line) < 100
            and not re.match(r'^\d+[-.)]', line)
            and not line.endswith(":")
        )

        # إذا كان إضافة السطر ستجعل الـ chunk أكبر من الحد
        if current_chunk and len(current_chunk) + len(line) + 1 > max_chunk_length:
            chunks.append(current_chunk.strip())
            current_chunk = ""

        # إضافة النص الحالي
        if current_chunk:
            current_chunk += " " + line
        else:
            current_chunk = line

    # إضافة آخر chunk
    if current_chunk:
        chunks.append(current_chunk.strip())

    return chunks


# إنشاء chunks لكل الوثائق
all_chunks = []

for doc in documents:

    chunks = create_chunks(doc["normalized_text"])

    for i, chunk in enumerate(chunks, start=1):

        all_chunks.append({
            "chunk_id": f'{doc["document_id"]}_CHUNK_{i:03d}',
            "document_id": doc["document_id"],
            "title": doc["title"],
            "url": doc["url"],
            "language": doc["language"],
            "text": chunk
        })

print("Chunking completed!")
print("Total chunks:", len(all_chunks))
```

```python
# تجهيز النص مع الحفاظ على الأسطر والعناوين

import re

def normalize_arabic_keep_lines(text):
    lines = text.splitlines()
    cleaned_lines = []

    for line in lines:
        line = line.strip()

        if not line:
            continue

        # توحيد أشكال الألف
        line = re.sub(r'[إأآٱ]', 'ا', line)

        # توحيد الألف المقصورة
        line = line.replace('ى', 'ي')

        # إزالة التشكيل
        line = re.sub(
            r'[\u0617-\u061A\u064B-\u065F\u0670\u06D6-\u06ED]',
            '',
            line
        )

        # إزالة التطويل
        line = line.replace('ـ', '')

        # توحيد المسافات داخل السطر فقط
        line = re.sub(r'[ \t]+', ' ', line).strip()

        if line:
            cleaned_lines.append(line)

    return "\n".join(cleaned_lines)


# تطبيق النسخة الجديدة على جميع الوثائق
for doc in documents:
    doc["normalized_text"] = normalize_arabic_keep_lines(
        doc["content"]
    )

print("Normalization with line preservation completed!")
```

```python
# Section-based Chunking

import re

def is_heading(line):
    line = line.strip()

    if not line:
        return False

    # العناوين عادة قصيرة
    if len(line) > 150:
        return False

    # استبعاد الأسطر التي تبدأ كقائمة
    if re.match(r'^\d+[\-.)]', line):
        return False

    # استبعاد الأسطر التي تبدو كروابط
    if line.startswith("http"):
        return False

    # إذا كان السطر سؤالاً فهو غالباً عنوان FAQ
    if line.endswith("؟"):
        return True

    # إذا كان السطر لا يحتوي على علامات ترقيم قوية
    punctuation = ".,:؛،!؟"

    if not any(char in line for char in punctuation):
        return True

    return False


# اختبار العناوين في أول وثيقة
doc = documents[0]

lines = doc["normalized_text"].splitlines()

headings = []

for line in lines:
    if is_heading(line):
        headings.append(line)

print("Number of detected headings:", len(headings))
print("\nFirst 30 detected headings:\n")

for heading in headings[:30]:
    print("-", heading)
```

```python
# إنشاء الـ Chunks من الأقسام

MAX_CHUNK_LENGTH = 1200
OVERLAP = 150

all_chunks = []


def split_long_text(text, max_length=1200, overlap=150):
    """
    تقسيم النص الطويل إلى أجزاء مع overlap بسيط.
    نحاول التقسيم عند نهاية الجمل قدر الإمكان.
    """

    if len(text) <= max_length:
        return [text.strip()]

    chunks = []
    start = 0

    while start < len(text):

        end = min(start + max_length, len(text))

        # البحث عن نهاية جملة قريبة من الحد الأقصى
        if end < len(text):
            possible_end = max(
                text.rfind("؟", start, end),
                text.rfind(".", start, end),
                text.rfind("؛", start, end)
            )

            if possible_end > start + 300:
                end = possible_end + 1

        chunk = text[start:end].strip()

        if chunk:
            chunks.append(chunk)

        # إذا وصلنا للنهاية
        if end >= len(text):
            break

        # Overlap
        start = max(end - overlap, start + 1)

    return chunks


for doc in documents:

    lines = [
        line.strip()
        for line in doc["normalized_text"].splitlines()
        if line.strip()
    ]

    current_section = "عام"
    current_text = []

    sections = []

    for line in lines:

        if is_heading(line):

            # حفظ القسم السابق
            if current_text:
                sections.append({
                    "section": current_section,
                    "text": " ".join(current_text).strip()
                })

            current_section = line
            current_text = []

        else:
            current_text.append(line)

    # حفظ آخر قسم
    if current_text:
        sections.append({
            "section": current_section,
            "text": " ".join(current_text).strip()
        })

    # تحويل الأقسام إلى Chunks
    for section in sections:

        section_name = section["section"]
        section_text = section["text"]

        # تجاهل القسم الخاص بكلمة السر
        if "كلمة السر" in section_name:
            continue

        # دمج العنوان مع المحتوى
        full_text = f"{section_name}: {section_text}"

        text_parts = split_long_text(
            full_text,
            max_length=MAX_CHUNK_LENGTH,
            overlap=OVERLAP
        )

        for part_number, chunk_text in enumerate(text_parts, start=1):

            all_chunks.append({
                "chunk_id": f'{doc["document_id"]}_CHUNK_{len(all_chunks) + 1:04d}',
                "document_id": doc["document_id"],
                "title": doc["title"],
                "section": section_name,
                "url": doc["url"],
                "language": doc["language"],
                "text": chunk_text
            })


print("Chunking completed!")
print("Total chunks:", len(all_chunks))
```

```python
# النسخة النهائية من اكتشاف العناوين

def is_heading(line):
    line = line.strip()

    if not line:
        return False

    # العناوين قصيرة نسبيًا
    if len(line) > 150:
        return False

    # استبعاد القوائم المرقمة
    if re.match(r'^\d+[\-.)]', line):
        return False

    # استبعاد الروابط
    if line.startswith("http"):
        return False

    # استبعاد Footer
    if "جميع الحقوق محفوظة" in line:
        return False

    # استبعاد الأسطر التي تنتهي بعلامات استمرار الجملة
    if line.endswith((":", "،", "؛", ",")):
        return False

    # أسئلة FAQ تعتبر عناوين
    if line.endswith("؟"):
        return True

    # عبارات تبدأ عادة كمحتوى وليس كعنوان
    sentence_starters = [
        "ويمكن",
        "كما يمكن",
        "كما يتم",
        "كما يحق",
        "ويتم",
        "ويحق",
        "وفي حال",
        "في حال",
        "وفي حالة",
        "في حالة",
        "علما",
        "حيث",
        "لذلك",
        "لذا",
        "اذا",
        "إذا",
        "من خلال",
        "وفقا",
        "حسب",
        "علي",
        "على",
        "ان",
        "إن",
        "وكلمة السر"
    ]

    for starter in sentence_starters:
        if line.startswith(starter + " "):
            return False

    # الكلمات القصيرة جدًا غالبًا ليست عناوين
    words = line.split()

    if len(words) <= 2:
        valid_short_headings = [
            "التاجيل",
            "تغيير التخصص",
            "نظام الدراسة",
            "لغة الدراسة",
            "مدة الدراسة"
        ]

        if line not in valid_short_headings:
            return False

    # إذا لم توجد علامات ترقيم، نعتبره Heading
    punctuation = ".,:؛،!؟"

    if not any(char in line for char in punctuation):
        return True

    return False


# اختبار العناوين النهائية
doc = documents[0]

headings = [
    line.strip()
    for line in doc["normalized_text"].splitlines()
    if is_heading(line)
]

print("Number of detected headings:", len(headings))

print("\nDetected headings:\n")

for heading in headings:
    print("-", heading)
```

```python
# إعادة إنشاء الـ Chunks باستخدام Heading Detection النهائي

MAX_CHUNK_LENGTH = 1200
OVERLAP = 150

all_chunks = []


def split_long_text(text, max_length=1200, overlap=150):
    """تقسيم النصوص الطويلة مع overlap بسيط."""

    if len(text) <= max_length:
        return [text.strip()]

    chunks = []
    start = 0

    while start < len(text):

        end = min(start + max_length, len(text))

        # محاولة إنهاء الـ chunk عند نهاية جملة
        if end < len(text):

            possible_end = max(
                text.rfind("؟", start, end),
                text.rfind(".", start, end),
                text.rfind("؛", start, end)
            )

            if possible_end > start + 300:
                end = possible_end + 1

        chunk = text[start:end].strip()

        if chunk:
            chunks.append(chunk)

        if end >= len(text):
            break

        # Overlap
        start = max(end - overlap, start + 1)

    return chunks


for doc in documents:

    lines = [
        line.strip()
        for line in doc["normalized_text"].splitlines()
        if line.strip()
    ]

    sections = []

    current_section = "عام"
    current_text = []

    for line in lines:

        if is_heading(line):

            # حفظ القسم السابق
            if current_text:
                sections.append({
                    "section": current_section,
                    "text": " ".join(current_text).strip()
                })

            current_section = line
            current_text = []

        else:
            current_text.append(line)

    # حفظ آخر قسم
    if current_text:
        sections.append({
            "section": current_section,
            "text": " ".join(current_text).strip()
        })


    # تحويل الأقسام إلى Chunks
    for section in sections:

        section_name = section["section"]
        section_text = section["text"]

        if not section_text:
            continue

        # حذف Footer إن وجد
        if "جميع الحقوق محفوظة" in section_text:
            section_text = section_text.split(
                "جميع الحقوق محفوظة"
            )[0].strip()

        if not section_text:
            continue

        # إضافة العنوان إلى النص
        full_text = f"{section_name}: {section_text}"

        text_parts = split_long_text(
            full_text,
            max_length=MAX_CHUNK_LENGTH,
            overlap=OVERLAP
        )

        for part_number, chunk_text in enumerate(
            text_parts,
            start=1
        ):

            chunk_id = (
                f'{doc["document_id"]}_'
                f'CHUNK_{len(all_chunks) + 1:04d}'
            )

            all_chunks.append({
                "chunk_id": chunk_id,
                "document_id": doc["document_id"],
                "title": doc["title"],
                "section": section_name,
                "url": doc["url"],
                "language": doc["language"],
                "text": chunk_text
            })


print("Final chunking completed!")
print("Total chunks:", len(all_chunks))
```

```python
#  BM25 Search Function

def bm25_search(query, top_k=5):
    
    # تجهيز الاستعلام
    query_tokens = tokenize_arabic(query)
    
    # حساب الدرجات
    scores = bm25.get_scores(query_tokens)
    
    # الحصول على أعلى النتائج
    top_indices = scores.argsort()[::-1][:top_k]
    
    results = []
    
    for rank, idx in enumerate(top_indices, start=1):
        
        chunk = chunks[idx]
        
        results.append({
            "rank": rank,
            "chunk_id": chunk["chunk_id"],
            "document_id": chunk["document_id"],
            "section": chunk["section"],
            "score": float(scores[idx]),
            "text": chunk["text"]
        })
    
    return results
```

```python
import math

def calculate_recall_at_k(results, relevant_ids, k):
    top_k_ids = [r["chunk_id"] for r in results[:k]]
    return 1 if any(chunk_id in relevant_ids for chunk_id in top_k_ids) else 0


def calculate_mrr(results, relevant_ids):
    for rank, result in enumerate(results, start=1):
        if result["chunk_id"] in relevant_ids:
            return 1 / rank
    return 0


def calculate_ndcg_at_k(results, relevant_ids, k):
    dcg = 0

    for rank, result in enumerate(results[:k], start=1):
        if result["chunk_id"] in relevant_ids:
            dcg += 1 / math.log2(rank + 1)

    ideal_relevant = min(len(relevant_ids), k)

    idcg = sum(
        1 / math.log2(rank + 1)
        for rank in range(1, ideal_relevant + 1)
    )

    if idcg == 0:
        return 0

    return dcg / idcg


# Calculate metrics for each query
bm25_metrics = []

for evaluation in bm25_evaluation_results:

    results = evaluation["results"]
    relevant_ids = evaluation["relevant_chunk_ids"]

    recall_5 = calculate_recall_at_k(
        results,
        relevant_ids,
        5
    )

    recall_10 = calculate_recall_at_k(
        results,
        relevant_ids,
        10
    )

    mrr = calculate_mrr(
        results,
        relevant_ids
    )

    ndcg_10 = calculate_ndcg_at_k(
        results,
        relevant_ids,
        10
    )

    bm25_metrics.append({
        "query_id": evaluation["query_id"],
        "Recall@5": recall_5,
        "Recall@10": recall_10,
        "MRR": mrr,
        "NDCG@10": ndcg_10
    })


print("BM25 metrics calculated successfully!")
print("\nResults:")
for metric in bm25_metrics:
    print(metric)
```

```python
# ==========================================
# Hybrid Retrieval: BM25 + BGE-M3
# Using Reciprocal Rank Fusion (RRF)
# ==========================================

def hybrid_search(query_embedding, query_text, top_k=10, rrf_k=60):
    
    # -------------------------
    # 1. BM25 ranking
    # -------------------------
    query_tokens = tokenize_arabic(query_text)
    bm25_scores = bm25.get_scores(query_tokens)
    
    bm25_indices = bm25_scores.argsort()[::-1]
    
    # -------------------------
    # 2. Dense ranking
    # -------------------------
    dense_scores = np.dot(chunk_embeddings, query_embedding)
    
    dense_indices = dense_scores.argsort()[::-1]
    
    # -------------------------
    # 3. RRF scores
    # -------------------------
    rrf_scores = np.zeros(len(chunks))
    
    for rank, idx in enumerate(bm25_indices, start=1):
        rrf_scores[idx] += 1 / (rrf_k + rank)
    
    for rank, idx in enumerate(dense_indices, start=1):
        rrf_scores[idx] += 1 / (rrf_k + rank)
    
    # -------------------------
    # 4. Get top-k
    # -------------------------
    top_indices = rrf_scores.argsort()[::-1][:top_k]
    
    results = []
    
    for rank, idx in enumerate(top_indices, start=1):
        chunk = chunks[idx]
        
        results.append({
            "rank": rank,
            "chunk_id": chunk["chunk_id"],
            "document_id": chunk["document_id"],
            "section": chunk["section"],
            "score": float(rrf_scores[idx]),
            "text": chunk["text"]
        })
    
    return results

print("Hybrid search function created successfully!")
```

```python
# ==========================================
# Run Hybrid Retrieval on all 50 queries
# ==========================================

hybrid_results = []

for i, item in enumerate(evaluation_queries_final):
    
    results = hybrid_search(
        query_embedding=query_embeddings[i],
        query_text=item["query"],
        top_k=10
    )
    
    hybrid_results.append({
        "query_id": item["query_id"],
        "query": item["query"],
        "relevant_chunk_ids": item["relevant_chunk_ids"],
        "results": results
    })

print("Hybrid retrieval completed!")
print("Number of queries:", len(hybrid_results))
```

```python
# ==========================================
# Evaluate Hybrid Retrieval
# ==========================================

hybrid_evaluation = []

for item in hybrid_results:
    
    results = item["results"]
    relevant_ids = item["relevant_chunk_ids"]
    
    hybrid_evaluation.append({
        "query_id": item["query_id"],
        "recall@5": calculate_recall_at_k(
            results, relevant_ids, 5
        ),
        "recall@10": calculate_recall_at_k(
            results, relevant_ids, 10
        ),
        "MRR": calculate_mrr(
            results, relevant_ids
        ),
        "NDCG@10": calculate_ndcg_at_k(
            results, relevant_ids, 10
        )
    })

hybrid_df = pd.DataFrame(hybrid_evaluation)

print("Hybrid evaluation completed!")
```

```python
def rerank_results(query, results):
    pairs = [
        [query, result["text"]]
        for result in results
    ]

    inputs = tokenizer(
        pairs,
        padding=True,
        truncation=True,
        return_tensors="pt"
    )

    with torch.no_grad():
        outputs = reranker_model(**inputs)

    scores = outputs.logits.view(-1).cpu().numpy()

    reranked = []

    for result, score in zip(results, scores):
        reranked.append({
            **result,
            "reranker_score": float(score)
        })

    reranked.sort(
        key=lambda x: x["reranker_score"],
        reverse=True
    )

    for rank, result in enumerate(reranked, start=1):
        result["rerank_rank"] = rank

    return reranked
```

```python
reranked_all = []

for i, item in enumerate(evaluation_queries_final):

    query = item["query"]

    # Hybrid retrieves Top-20 candidates
    hybrid_results = hybrid_search(
        query_embedding=query_embeddings[i],
        query_text=query,
        top_k=20
    )

    # Rerank the Top-20
    reranked_results = rerank_results(
        query=query,
        results=hybrid_results
    )

    # Keep Top-10 after reranking
    top_10 = reranked_results[:10]

    reranked_all.append({
        "query_id": item["query_id"],
        "query": query,
        "relevant_chunk_ids": item["relevant_chunk_ids"],
        "results": top_10
    })

print("Reranking completed for", len(reranked_all), "queries.")
```
